```python

import pandas as pd
from google.colab import files

# Task 2: Data Cleaning and Schema Transformation

# 1. Upload your existing CSV file
uploaded = files.upload()

# Get the uploaded CSV filename
file_name = next(iter(uploaded))

# 2. Read the uploaded CSV file
df = pd.read_csv(file_name)

print("RAW DATA")
print(df.head())

# 3. Clean column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
)

# 4. Clean text columns
for col in df.select_dtypes(include=["object"]).columns:
    df[col] = df[col].str.strip()

# Standardize capitalization for text values
for col in df.select_dtypes(include=["object"]).columns:
    df[col] = df[col].str.title()

# 5. Remove duplicate rows
df = df.drop_duplicates()

# 6. Handle missing values
# Remove rows where all values are missing
df = df.dropna(how="all")

# Fill missing numeric values with the column median
for col in df.select_dtypes(include="number").columns:
    if df[col].isna().any():
        df[col] = df[col].fillna(df[col].median())

# Fill missing text values with "Unknown"
for col in df.select_dtypes(include=["object"]).columns:
    df[col] = df[col].fillna("Unknown")

# 7. Validate the cleaned data
assert not df.duplicated().any()
assert not df.isna().any().any()

print("\nCLEANED DATA")
print(df)

print("\nDATA TYPES")
print(df.dtypes)

# 8. Save cleaned data
output_file = "cleaned_customers.csv"
df.to_csv(output_file, index=False)

print("\nCleaning completed successfully!")
print("Saved file:", output_file)

# 9. Download the cleaned CSV
files.download(output_file)
```
